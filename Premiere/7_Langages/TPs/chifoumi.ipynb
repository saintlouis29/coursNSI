# TP Chifoumi

![](chifoumi.jpg)


## Règles

L’ordinateur et vous allez choisir entre *pierre*, *feuille* et *ciseaux*.

Le gagnant est désigné selon ces règles:

- La pierre brise les ciseaux;
- Le papier couvre la pierre;
- Les ciseaux coupent le papier;


## Consignes

Nous allons utiliser des **fonctions** afin de mieux **structurer** notre programme.

### Les fonctions "intelligentes" de notre programme

Nous allons utiliser ces conventions:

Pour le **choix**:
- 0 : Pierre
- 1 : Feuille
- 2 : Ciseaux

Pour le **verdict**:
- -1 : Perdu
- 0 : Egalité
- 1 : Gagné

Ecrire la fonction **verdict_chifoumi** qui retourne le verdict du point de vue du joueur1.
  - **Entrées** :
      - joueur1 : int (0, 1 ou 2) 
      - joueur2 : int (0, 1 ou 2) 
  - **Retourne** un entier:
      - 0 si il y a égalité
      - -1 si le joueur1 a perdu
      - 1 si le joueur1 a gagné


Ecrire la fonction **correspondance_verdict_label** qui retourne le verdict sous forme de chaine de caractère.
  - **Entrée** :
      - verdict : int (0, 1 ou 2) 
  - **Retourne** une chaîne de caractère:
      - "Egalité" si le verdict était 0
      - "Perdu" si le verdict était -1
      - "Gagné" si le verdict était 1


Ecrire la fonction **correspondance_choix_label** qui retourne le choix sous forme de chaine de caractère.
  - **Entrée** :
      - choix : int (0, 1 ou 2) 
  - **Retourne** une chaîne de caractère:
      - "P" si le choix était 0
      - "F" si le choix était 1
      - "C" si le choix était 2


Ecrire la fonction **correspondance_choix_entier** qui retourne le choix sous forme d'un entier.
  - **Entrée** :
      - choix : une chaîne de caractère ("P", "F" ou "C") 
  - **Retourne** un entier:
      - 0 si le choix était "P"
      - 1 si le choix était "F"
      - 2 si le choix était "C"

Ecrire la fonction **choix_aleatoire** qui retourne un choix aléatoire parmi 0, 1 ou 2.
  - **Entrée** : aucune
  - **Retourne** un entier: 0, 1 ou 2

Vous pourrez utilisez la fonction [randint](https://www.w3schools.com/python/ref_random_randint.asp) du module *random*.


### Les fonctions d'interface du programme

Il ne reste plus qu'à écrire les fonctions d'interface qui permettent:
- à l'utilisateur d'**entrer** son choix (*input*)
- au programme d'**afficher** les resultats (*print*)
 
Ecrire une fonction **interface_partie** qui demande à l'utilisateur de choisir "P", "F", "C". L'ordinateur choisit également et le programme et affiche son choix. Finalement on affiche "Egalité", "Perdu" ou "Gagné".
  - **Entrées** : aucune
  - **Retourne** : Rien

Ecrire une fonction **interface_jeu** qui permet de faire des parties de chifoumi jusqu'à ce que l'utilisateur décide de quitter le programme (q).
  - **Entrées** : aucune
  - **Retourne** : Rien

### Améliorations possibles:

- Si l'utilisateur entre autre chose que "P", "p", "F", "f", "C", "c", un message affiche "Veuillez choisir P, F ou C"
