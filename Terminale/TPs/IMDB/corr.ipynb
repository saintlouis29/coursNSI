# Corrigé

## Création des clés étrangères

Clé étrangère n° 1 : les participations vers les œuvres

```sql
ALTER TABLE title_principals
ADD CONSTRAINT fk_principals_title
FOREIGN KEY (tconst)
REFERENCES title_basics(tconst);
```

Clé étrangère n° 2 : les participations vers les personnes

```sql
ALTER TABLE title_principals
ADD CONSTRAINT fk_principals_name
FOREIGN KEY (nconst)
REFERENCES name_basics(nconst);
```
Clé étrangère n° 3 : les notes vers les œuvres

```sql
ALTER TABLE title_ratings
ADD CONSTRAINT fk_ratings_title
FOREIGN KEY (tconst)
REFERENCES title_basics(tconst);
```
